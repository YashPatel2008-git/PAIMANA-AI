```python
import pandas as pd 
import numpy as np 
from pathlib import Path 
# Project paths 
BASE_DIR = Path("..") 
PROCESSED_DIR = BASE_DIR / "data" / "processed" 
REPORT_DIR = BASE_DIR / "reports" 
REPORT_DIR.mkdir(parents=True, exist_ok=True) 
print("Processed folder:", PROCESSED_DIR.resolve())
print("Reports folder:", REPORT_DIR.resolve()) 
# Load available processed datasets 
cost_file = PROCESSED_DIR / "cost_model_dataset.csv" 
time_file = PROCESSED_DIR / "time_model_dataset.csv" 
master_file = PROCESSED_DIR / "project_master.csv" 
cost_df = pd.read_csv(cost_file) 
time_df = pd.read_csv(time_file) 
master_df = pd.read_csv(master_file) 
print("\nDatasets loaded successfully.") 
print("Cost dataset shape:", cost_df.shape) 
print("Time dataset shape:", time_df.shape) 
print("Project master shape:", master_df.shape)
```

```python
# Columns that should be used as model INPUTS 
predictor_columns = [    
    "PROJECT_ID",    
    "MINISTRY",    
    "SECTOR",    
    "STATE",    
    "AGENCY",   
    "ORIGINAL_COST_RS_CRORE",    
    "CUMULATIVE_EXPENDITURE_RS_CRORE",    
    "PHYSICAL_PROGRESS_PERCENT",    
    "START_DATE",    
    "ORIGINAL_TARGET_DOC",    
    "PROGRESS_CHANGE",    
    "EXPENDITURE_CHANGE" ] 
# Keep only columns that actually exist 
available_predictors = [    
    col for col in predictor_columns    
    if col in master_df.columns ] 
ml_predictors = master_df[available_predictors].copy() 
# Save 
predictor_output = PROCESSED_DIR / "member1_ml_predictors.csv" 
ml_predictors.to_csv(predictor_output, index=False) 
print("ML predictor dataset created.") 
print("Shape:", ml_predictors.shape) 
print("Saved to:", predictor_output) 
print("\nColumns:") 
print(ml_predictors.columns.tolist())
```

```python
# Target variables for prediction 
target_columns = [    
    "PROJECT_ID",    
    "REPORT_MONTH",    
    "COST_OVERRUN_PCT",    
    "SCHEDULE_CHANGE_DAYS",    
    "DELAY_INDICATOR" ] 
available_targets = [    
    col for col in target_columns    
    if col in master_df.columns ] 
ml_targets = master_df[available_targets].copy() 

# Save target dataset 
target_output = PROCESSED_DIR / "member1_ml_targets.csv" 
ml_targets.to_csv(target_output, index=False) 
print("ML target dataset created.") 
print("Shape:", ml_targets.shape) 
print("Saved to:", target_output) 
print("\nTarget columns:") 
print(ml_targets.columns.tolist())
```

```python
# Missing-value analysis 
quality_df = pd.DataFrame({    
    "COLUMN": master_df.columns,
    "DATA_TYPE": master_df.dtypes.astype(str).values,    
    "MISSING_COUNT": master_df.isna().sum().values }) 
quality_df["TOTAL_ROWS"] = len(master_df) 
quality_df["MISSING_PERCENT"] = (    
    quality_df["MISSING_COUNT"] /    
    quality_df["TOTAL_ROWS"] * 100 ).round(2) 
quality_df["STATUS"] = np.where(    
    quality_df["MISSING_COUNT"] == 0,    
    "NO MISSING VALUES",    
    "MISSING VALUES PRESENT" ) 
quality_output = REPORT_DIR / "member1_data_quality.csv" 
quality_df.to_csv(    quality_output,    index=False ) 

print("Data-quality report created.") 
print("Saved to:", quality_output) 
display(quality_df)
```

```python
# Duplicate validation 
exact_duplicates = master_df.duplicated().sum() 
if "PROJECT_ID" in master_df.columns and "REPORT_MONTH" in master_df.columns:    
    project_month_duplicates = master_df.duplicated(        
        subset=["PROJECT_ID", "REPORT_MONTH"]    ).sum() 
else:    project_month_duplicates = "Columns unavailable"
print("========== DUPLICATE CHECK ==========") 
print("Exact duplicate rows:", exact_duplicates) 
print("DuplicatePROJECT ID+REPORT MONTH:",project_month_duplicates)
```
