```python
import pandas as pd 
import numpy as np 
from pathlib import Path 
print("Libraries loaded successfully!")
```

```python
from pathlib import Path 
raw_folder = Path("../data/raw") 
excel_files = list(raw_folder.glob("*.xlsx"))
print("Excel files found:")
for file in excel_files:    
    print("→", file.name) 
    print("\nTotal files:", len(excel_files))
```

```python
header=3
all_months = [] 
for file in excel_files:        
    # Read the Project Monitoring sheet    
    df = pd.read_excel(        
        file,        
        sheet_name="Table 6 - All Ongoing",        
        header=3
    )       
    # Get month from filename    
    filename = file.stem        
    if "April" in filename:        
        month = "APRIL"    
    elif "May" in filename:        
        month = "MAY"    
    elif "June" in filename:        
        month = "JUNE"    
    elif "July" in filename:        
        month = "JULY"    
    else:        
        month = "UNKNOWN"        
        
    # Add month to every row    
    df["REPORT MONTH"] = month        
    # Add source filename    
    df["SOURCE FILE"] = file.name        
    # Store the dataframe    
    all_months.append(df)        
    
    print(f"{month}: {len(df)} rows loaded")
```

```python
combined = pd.concat(all_months,ignore_index=True ) 
print("Combined dataset created!") 
print("Rows:", len(combined)) 
print("Columns:", len(combined.columns))
```

```python
monthly_project_data = combined.copy() 
monthly_project_data = monthly_project_data.sort_values(
    by=["PROJECT CODE(S)","REPORT MONTH"]) 
print(monthly_project_data.head())
```

```python
month_order = [    "APRIL",    "MAY",    "JUNE",    "JULY" ] 
monthly_project_data["REPORT MONTH"] = pd.Categorical(    monthly_project_data["REPORT MONTH"], categories=month_order,    ordered=True ) 
monthly_project_data = monthly_project_data.sort_values(    ["PROJECT CODE(S)","REPORT MONTH"] )
```

```python
output_file = "../data/cleaned/paimana_all_months_combined.csv" 
monthly_project_data.to_csv(    output_file,    index=False ) 
print("Saved successfully:") 
print(output_file)
```
