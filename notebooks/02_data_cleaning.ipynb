```python
import pandas as pd 
import numpy as np 
from pathlib import Path 
print("Libraries loaded successfully")
```

```python
file_path = "../data/cleaned/paimana_all_months_combined.csv" 
df = pd.read_csv(file_path) 
print("Dataset loaded successfully") 
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
pd.set_option("display.max_columns", None) 
df.head()
```

```python
print("All columns:\n") 
for i, column in enumerate(df.columns, 1):    
    print(i, "→", column)
```

```python
df.columns = (    
    df.columns    
    .str.strip()    
    .str.upper()    
    .str.replace(" ", "_")    
    .str.replace("/", "_", regex=False)    
    .str.replace("(", "", regex=False)    
    .str.replace(")", "", regex=False)    
    .str.replace("%", "PCT", regex=False)    
    .str.replace(".", "", regex=False) 
)
```

```python
missing = pd.DataFrame({    
    "Missing_Count": df.isna().sum(),    
    "Missing_Percentage": (        
        df.isna().mean() * 100 )
        .round(2) }) 
missing = missing.sort_values(    "Missing_Percentage",    ascending=False ) 
missing
```

```python
duplicate_rows = df.duplicated().sum() 
print("Exact duplicate rows:", duplicate_rows)
```

```python
date_columns = [    
    col for col in df.columns    
    if "DATE" in col or "DOC" in col ] 
print(date_columns)
```

```python
for col in date_columns:    
    df[col] = pd.to_datetime(        
        df[col],
         errors="coerce"    
    ) 
print("Date conversion completed")
```

```python
numeric_keywords = [    "COST",    "EXPENDITURE",    "PROGRESS" ] 
numeric_columns = [    
    col for col in df.columns    
    if any(word in col for word in numeric_keywords) ] 
print(numeric_columns)
```

```python
for col in numeric_columns:    
    df[col] = (        
        df[col]        
        .astype(str)        
        .str.replace(",", "", regex=False)        
        .str.replace("₹", "", regex=False)        
        .str.strip()    )        
        
    df[col] = pd.to_numeric(
         df[col],        
         errors="coerce"    ) 
         
print("Numerical conversion completed")
```

```python
df[numeric_columns].describe().T
```

```python
progress_col = "PHYSICAL_PROGRESS_PCT"
print("Minimum:", df[progress_col].min()) 
print("Maximum:", df[progress_col].max())
print("Median:", df[progress_col].median())
```

```python
# part B 
df["PHYSICAL_PROGRESS_PERCENT"] = (df[progress_col] * 100)
df[    [        progress_col,        "PHYSICAL_PROGRESS_PERCENT"    ] ].head(20)
```

```python
# PART C - STANDARDIZE REPORT MONTH

df["REPORT_MONTH"] = (
    df["REPORT_MONTH"] 
    .astype(str)
    .str.strip()
    .str.upper()
)

# Correct month order
month_order = [
    "APRIL",
    "MAY",
    "JUNE",
    "JULY"
]

df["REPORT_MONTH"] = pd.Categorical(
    df["REPORT_MONTH"],
    categories=month_order,
    ordered=True
)

print("Report month standardized successfully!")
print(df["REPORT_MONTH"].value_counts())
```

```python
# PART d 

df = df.sort_values(    [PROJECT_ID, "REPORT_MONTH"] ) 
df.head(20)
```

```python
project_month_count = (    
    df.groupby(PROJECT_ID)["REPORT_MONTH"]    
    .nunique()    
    .sort_values(ascending=False)
)
print(project_month_count.value_counts())
```

```python
projects_4_months = project_month_count[    
    project_month_count == 4 ] 
print(    "Projects appearing in all 4 months:",    len(projects_4_months) )
```

```python
output_path = "../data/cleaned/paimana_cleaned.csv" 
df.to_csv(    output_path,    index=False ) 
print("Cleaned dataset saved to:") 
print(output_path)
```
